```python
import gradio as gr
```

```python
def gradio_interface(img):
  image = new_model.predict(img)
  image = image[0].plot()
  return image
```

```python
result=gr.Interface(fn=gradio_interface,inputs=gr.Image("numpy"),outputs=gr.Image("numpy"))
```

```python
result.launch()
```
